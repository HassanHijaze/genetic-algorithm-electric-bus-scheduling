```python
from dataclasses import dataclass
from typing import List


# A chain is an ordered sequence of service-trip IDs.
Chain = List[int]

# A block is an energy-feasible part of a chain.
# It starts from the depot, serves one or more trips,
# and can return to the depot with the required reserve.
Block = List[int]


@dataclass
class EnergyConfig:
    battery_capacity: float
    min_soc_reserve: float
    charge_rate_wh_per_hour: float
    allow_equal_time: bool = True


@dataclass
class MergeCandidate:
    from_block: int
    to_block: int
    arrive_depot_time: int
    latest_depart_time: int
    required_charge_minutes: int
    depot_id: int

    # SoC at depot after block i returns.
    start_energy: float

    # SoC after charging during the dwell before block j.
    end_energy: float


@dataclass
class DwellEvent:
    from_block: int
    to_block: int
    depot_id: int
    start_time: int
    end_time: int
    start_energy: float
    end_energy: float
```
